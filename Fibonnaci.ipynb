```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)


def fibonacci_loop(count):
    first = 0
    second = 1

    for i in range(count):
        print(first)
        first, second = second, first + second


def recursive_sort(numbers, start=0):
    if start >= len(numbers) - 1:
        return numbers

    smallest = start

    for i in range(start + 1, len(numbers)):
        if numbers[i] < numbers[smallest]:
            smallest = i

    numbers[start], numbers[smallest] = numbers[smallest], numbers[start]

    return recursive_sort(numbers, start + 1)


print("Recursive Fibonacci:", fibonacci(6))

print("Fibonacci using a for-loop:")
fibonacci_loop(7)

numbers = [4, 2, 5, 1]
print("Sorted list:", recursive_sort(numbers))
```
